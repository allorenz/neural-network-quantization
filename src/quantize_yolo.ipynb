```python
import tensorflow as tf

# Load the SavedModel
converter = tf.lite.TFLiteConverter.from_saved_model("yolo_saved_model")

# Perform quantization (you can choose the quantization method)
converter.optimizations = [tf.lite.Optimize.DEFAULT]  # Dynamic range quantization
# For other quantization options, e.g. full integer quantization, you need to set additional settings.

# Convert to TFLite model
tflite_model = converter.convert()

# Save the TFLite model
with open("yolo11n_quantized.tflite", "wb") as f:
    f.write(tflite_model)
```

```python
import torch
from ultralytics import YOLO

# Load your YOLO model
model = YOLO("../weights/yolo11n.pt")

# Define the input shape for ONNX export (adjust as per your model input shape)
dummy_input = torch.randn(1, 3, 640, 640)

# Export the model to ONNX
model.model.eval()
torch.onnx.export(
    model.model, 
    dummy_input, 
    "yolo11n.onnx", 
    opset_version=12,  # Specify ONNX version here
    input_names=["input"], 
    output_names=["output"]
)
```

```python
img_path = "../data/car.jpeg"
results = model.predict(img_path)

for result in results:
    # result.boxes.cls contains the class IDs (as tensor)
    class_ids = result.boxes.cls
    identified_objects = [model.names.get(int(class_id)) for class_id in class_ids]
    
    print(identified_objects)
```
